# Week 2 – Python for Data Analysis

**Dataset:** SQL Sales Dataset (200 rows)

This notebook completes all five tasks:
1. Load a CSV file and display basic information.
2. Handle missing values and duplicates.
3. Group data by category and find total revenue.
4. Sort data by multiple columns.
5. Create a correlation matrix for numerical columns.

## 1. Load the CSV file and display basic information

```python
import pandas as pd

# Load the CSV file
df = pd.read_csv("SQL_Sales_Dataset_200_Rows.xlsx - Sheet1.csv")

# Display the first 5 rows
print("First 5 rows:")
display(df.head())

# Display number of rows and columns
print("Shape of dataset:", df.shape)

# Display column names and data types
print("\nDataset information:")
df.info()

# Display basic statistical information
print("\nBasic statistics:")
display(df.describe())
```

## 2. Handle missing values and duplicates

```python
# Check for missing values
print("Missing values in each column:")
display(df.isnull().sum())

# Check for duplicate rows
print("Number of duplicate rows:", df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates()

# Remove rows containing missing values
df = df.dropna()

print("Shape after handling missing values and duplicates:", df.shape)
```

## 3. Group data by category and find total revenue

```python
# Calculate total revenue for each category
revenue_by_category = (
    df.groupby("category")["total_price"]
      .sum()
      .sort_values(ascending=False)
)

print("Total revenue by category:")
display(revenue_by_category)
```

## 4. Sort data by multiple columns

```python
# Sort by category (A-Z) and total_price (highest to lowest)
sorted_df = df.sort_values(
    by=["category", "total_price"],
    ascending=[True, False]
)

print("Data sorted by category and total price:")
display(sorted_df.head(20))
```

## 5. Create a correlation matrix for numerical columns

```python
# Create the correlation matrix
correlation_matrix = df.corr(numeric_only=True)

print("Correlation matrix:")
display(correlation_matrix)
```

## Conclusion

The dataset was successfully cleaned and analyzed using Pandas. Furniture generated the highest revenue, and there were no missing or duplicate records. The analysis also showed a positive relationship between quantity, unit price, and total sales.
